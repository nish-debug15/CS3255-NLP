```python
import re

text = """UNBELIEVABLY... the driver, Alex, arrivED on 12/05/2024!
He was super helpful. However, the ride cost $45.50 which is pricey.
Contact support@taxi-tech.com or call (555) 019-2834 for refunds.
I was unhappy with the wait time though."""

print(text)
```

## Part A: Theoretical Analysis

**Sentence:** “The driver arrived on time.”

1. **Morphology:**
   The word **“arrived”** can be divided into the root **“arrive”** and the suffix **“-ed”**. The suffix **“-ed”** indicates that the action happened in the **past tense**.

2. **Lexical Analysis:**

   * **driver** → Noun (N)
   * **arrived** → Verb (V)

3. **Syntax:**

   * **Noun Phrase (NP):** “The driver”
   * **Verb Phrase (VP):** “arrived on time”

4. **Semantics:**
   Here, **“ride”** literally means a **journey or trip taken in a vehicle**, specifically a taxi ride.

5. **Pragmatics:**
   If the sentence “The driver arrived on time” appears in a **1-star review**, the intended meaning may be **sarcastic**, especially if the rest of the review describes a bad experience. The implied intent is to express **dissatisfaction or criticism**, rather than genuine praise.


```python
# Normalize to lowercase
normalized_text = text.lower()

# Remove unwanted punctuation while preserving:
# letters, numbers, spaces, $, ., parentheses, and hyphens
cleaned_text = re.sub(r"[^a-z0-9\s$()./@_-]", "", normalized_text)

print("Normalized Text:")
print(normalized_text)

print("\nCleaned Text:")
print(cleaned_text)
```

```python
# 1. Date: DD/MM/YYYY
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", cleaned_text)

# 2. Price: $ followed by digits and optional decimal digits
prices = re.findall(r"\$\d+(?:\.\d{2})?", cleaned_text)

# 3. Email Address
emails = re.findall(r"\b[\w.-]+@[\w.-]+\.\w+\b", cleaned_text)

# 4. Phone Number: (555) 019-2834
phones = re.findall(r"\(\d{3}\)\s?\d{3}-\d{4}", cleaned_text)

print("Dates found:", dates)
print("Prices found:", prices)
print("Emails found:", emails)
print("Phone numbers found:", phones)
```

```python
# 1. Agent / Doer: word ending in -er
agents = re.findall(r"\b\w+er\b", cleaned_text)

# 2. Action / Past tense: word ending in -ed
past_tense = re.findall(r"\b\w+ed\b", cleaned_text)

# 3. Negation: word beginning with un- or dis-
negations = re.findall(r"\b(?:un|dis)\w+\b", cleaned_text)

print("Agent words (-er):", agents)
print("Past tense words (-ed):", past_tense)
print("Negation words (un-/dis-):", negations)
```

```python
agents = re.findall(r"\b\w+(?<!sup)er\b", cleaned_text)
```

```python
# Matches both:
# (555) 019-2834
# 555-019-2834

phones_bonus = re.findall(r"\(?\d{3}\)?\s?\d{3}-\d{4}", cleaned_text)

print("Phone numbers found:", phones_bonus)
```
